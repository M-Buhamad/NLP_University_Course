```python
# Task 1: Cosine Similarity
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Define the sentences
sentences_t1 = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first document?'
]

# Apply TF-IDF and calculate cosine similarity
tfidf_vectorizer_t1 = TfidfVectorizer()
vector_matrix_t1 = tfidf_vectorizer_t1.fit_transform(sentences_t1)
cosine_matrix = cosine_similarity(vector_matrix_t1)

# Display results in a DataFrame
df_cosine = pd.DataFrame(cosine_matrix, index=sentences_t1, columns=sentences_t1)
df_cosine
```

```python
# Task 2: TF-IDF
# Define the sentences
sentences_t2 = [
    'data science is one of the most important fields of science',
    'this is one of the best data science courses',
    'data scientists analyze data'
]

# Apply TF-IDF
tfidf_t2 = TfidfVectorizer()
result_t2 = tfidf_t2.fit_transform(sentences_t2)

# Display the important words (features) with their scores
df_tfidf = pd.DataFrame(result_t2.toarray(), columns=tfidf_t2.get_feature_names_out())
df_tfidf
```

```python
!pip install --upgrade gensim
# Task 3: Word2vec - Preprocessing and Model Creation
import re
import pandas as pd
import gensim
from nltk.tokenize import word_tokenize
import nltk

# Ensure punkt is downloaded for tokenization
nltk.download('punkt')

# Text cleaning function
def clean_text(text):
    if not isinstance(text, str):
        return ""
    text = text.lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

# Load dataset and apply preprocessing
df_simpsons = pd.read_csv('simpsons_script_lines.csv', engine='python', on_bad_lines='skip')
df_simpsons['cleaned_spoken_words'] = df_simpsons['spoken_words'].apply(clean_text)

# Tokenize the cleaned text
tokens = [word_tokenize(sentence) for sentence in df_simpsons['cleaned_spoken_words'] if sentence]

# Create the skip-gram Word2Vec model
Skip_gram_model = gensim.models.Word2Vec(tokens, min_count=1, vector_size=100, window=5, sg=1)
```

```python
# Task 4: Find similar words using wv.most_similar()
# Find words similar to "homer", "marge", and "bart"

print("Words similar to 'homer':\n", Skip_gram_model.wv.most_similar('homer'))
print("\nWords similar to 'marge':\n", Skip_gram_model.wv.most_similar('marge'))
print("\nWords similar to 'bart':\n", Skip_gram_model.wv.most_similar('bart'))
```
