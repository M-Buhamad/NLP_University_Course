```python
# Import spaCy and load the English model
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
# Task 1: POS tagging using spaCy

sen = nlp("I like watching football matches. I do not know how to play it though")

# Find the word "play"
for i, word in enumerate(sen):
    if word.text == "play":
        print("POS tag:", sen[i].pos_)
        print("Fine-grained POS tag:", sen[i].tag_)
        print("Tag meaning:", spacy.explain(sen[i].tag_))

# Print information for all words
for word in sen:
    print(f'{word.text:{12}} {word.pos_:{10}} {word.tag_:{8}} {spacy.explain(word.tag_)}')
```

```python
# Task 2: Compare the POS tags of Google and table

sentences = [
    "Did you search on Google?",
    "Let us google it.",
    "He has placed the food on the table.",
    "She plans to table a motion for debate in Parliament."
]

for sentence in sentences:
    sen = nlp(sentence)
    print("\nSentence:", sentence)

    for word in sen:
        if word.text.lower() in ["google", "table"]:
            print(
                word.text,
                word.pos_,
                word.tag_,
                spacy.explain(word.tag_)
            )
```

```python
# Task 3: Visualize POS and dependency parsing

from spacy import displacy

doc = nlp("I like to play football. I hated it in my childhood though")

displacy.render(
    doc,
    style='dep',
    jupyter=True,
    options={'distance': 110}
)
```
