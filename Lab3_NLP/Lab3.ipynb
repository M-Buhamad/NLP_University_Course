```python
!pip install emoji
import pandas as pd
import re
import nltk
import emoji

from nltk.tokenize import word_tokenize
from nltk.lm.preprocessing import padded_everygram_pipeline
from nltk.lm import MLE
```

```python
# Load the dataset
df = pd.read_csv("Customer_Review.csv")

df.head()
```

```python
# Check the dataset
df.info()
```

```python
# Select the reviews
texts = df["Label"].dropna().astype(str)
```

```python
# Remove hashtags, mentions, websites, RT and emojis
def clean_text(text):
    text = re.sub(r"#\w+", "", text)
    text = re.sub(r"@\w+", "", text)
    text = re.sub(r"http\S+|www\S+", "", text)
    text = re.sub(r"\bRT\b", "", text)
    text = emoji.replace_emoji(text, replace="")
    return text.strip()
```

```python
# Clean the reviews
cleaned_texts = texts.apply(clean_text)

cleaned_texts.head()
```

```python
# Tokenize the reviews
tokenized_texts = [
    word_tokenize(text.lower())
    for text in cleaned_texts
]
```

```python
# Create a bigram language model
train, vocab = padded_everygram_pipeline(
    2,
    tokenized_texts
)
```

```python
# Train the Maximum Likelihood Model
lm = MLE(2)
lm.fit(train, vocab)
```

```python
# Evaluate the model
print("Vocabulary size:", len(lm.vocab))
```

```python
# Calculate the probability of the bigram
print("Probability of 'is' after 'pakistan':")
print(lm.score("is", ["pakistan"]))
```

```python
# Calculate perplexity
test = [("pakistan", "is")]

print("Perplexity:")
print(lm.perplexity(test))
```
