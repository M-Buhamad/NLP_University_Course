```python
import pandas as pd
import re
from collections import Counter
```

```python
# Load the dataset
df = pd.read_csv("Customer_Review.csv")

df.head()
```

```python
# Check the dataset
df.info()
```

```python
# Show the column names
print(df.columns)
```

```python
# Find all hashtags in the reviews
hashtags = []

for text in df["Label"].dropna():
    hashtags.extend(re.findall(r"#\w+", text))

print("Total hashtags:", len(hashtags))
```

```python
# Count the hashtags
hashtag_counts = Counter(hashtags)

# Show the top 10 hashtags
print(hashtag_counts.most_common(10))
```

```python
import re

text = "This year is 2021"

# Create a pattern for digits
pattern = re.compile(r"\d+")

print(type(pattern))

updated_text = pattern.sub("2022", text)

print(updated_text)
```

```python
import re

text = "a 11 b 2 3 c 4"

# Split the text at digits
result = re.split(r"\d+", text)

print(result)
```
